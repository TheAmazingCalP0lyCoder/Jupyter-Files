6.  While my microphone was measuring my speaker signal, I measured voltage ($V$, sensed by the microphone) as a function of distance between the microphone and speaker ($R$).  See my data below:

```python
R = [0, 2, 4, 6, 8, 10, 12, 14, 16] #in cm
V = [1090, 510, 385, 290, 220, 180, 150, 125, 115] #in mV
```

How does measured voltage vary with distance?  Include plot(s) and/or fit(s) to support your conclusion.

<span style = 'background-color:yellow'>Answer to Problem 6:</span>  <br>
Voltage measured from the microphone has some inverse exponential relationship with the distance of the microphone. <br>
This can be seen when plotting the data and finding an exponential line of best fit. By plotting the data (see below) it plots a line very similar to that of an inverse exponential equation, as the measured voltage drops very quickly at first, then flattens out as distance increases. This is further supported with an exponential line of best fit plotted in the same graph. Only a first-order line, it follows very well to most of the data points after the first plot point. 
This makes sense physically, as there is most likely enviornmental frequency interferance caused by other sounds around the epxeriment. The environmental frequency makes substantial interference if the microphone is not close to the measurement source, so the microphone as it is pulled away measures more of the environmental "voltage" caused by the frequencies, rather than the speaker.

```python
a, b = np.polyfit(R,np.log(V), 1)

print(a)
print(b)
```

```python
b = np.exp(b)
e_fit = b*np.exp(a*R)
```

```python
plt.plot(R,V, 'bo')
plt.plot(R, e_fit, 'r--')
plt.grid(True)
plt.xlabel('Distance (cm)')
plt.ylabel('Voltage Measured (mV)')
plt.axis([0,17, 0, 1100])
plt.savefig('Voltage_Microphone_Distance_efit_12Feb2026.png')
```
