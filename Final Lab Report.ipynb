### Goal/Background

The goal is to use a potentiometer to determine how many of a series of LED's are turned on.<br>
<br>
This was partly motivated by videos I've seen online of people using various devices to control sets of lights, combined with a video of someone using their laundry unit to play the Tokyo Drift opening riff. When I first heard the potentiometer described I thought of using it to control LED lights, and that captivated me. 

### Methods

First, I created a simple potentiometer circuit voltage divider with a 1k resistor. I measured the voltage drop across the potentiometer, and converted that to see the change in resistance of the potentiometer over time.<br>
<br>
Next I built the first comparator circuit. I tested this circuit separately, made sure it gave a voltage jump at the set voltage, and then built two more.<br>
<br>
After the first comparator was successfully tested, I connected it to the voltage out of the potentiometer voltage divider. Skipping copious amounts of fussing a circuitry frustration, I was successful with having the three comparators set in parallel with each other, each comparator receiving the same voltage output from the voltage divider, but with individually set voltage points. Adding LEDs on the end of each comparator and testing by changing the resistance of the potentiometer resulted in each diode being individually turned on until all three were lit, then turning the potentiometer back to turn off each LED one by one.

### Conclusion

Looking at the results in the last plot, it can be seen that my goal was achieved. By changing the voltage drop across the potentiometer (by twisting its dial and changing its resistance value), I was able to turn on the LEDs one by one until all three were lit. Previous plots before the last one show the parts of the greater circuit behaving according to my desires to produce the final result. The resistance of the potentiometer for when each LED turned on was also calculated.<br>
<br>
The biggest possible source of error would be the comparators. It can be seen in the last couple of plots that the LEDs were receiving voltage even though the comparators were set to off via their set voltage points. This did not end up affecting the results of the LEDs turning on, but it is something to point out, as this means the comparator could have a noticeable discrepancy between its theoretical set voltage determined by its adjacent voltage divider, and the actual voltage point at which it let voltage flow freely through. <br>
<br>
