## Part 1:  Introduction to Fast Fourier Transforms (FFTs)

#### Oscillatory behavior

Many physical phenomena involve oscillatory behavior: vibrations of air molecules as sound travels across the room, or the physical motion of a pendulum as it swings back and forth.

It is common to measure and characterize these oscillations **as a function of time** and typically the quantity of interest (sound pressure or pendulum angle) can be expressed as:
$$y(t)=A\sin(\omega t+\phi)$$
where $y$ is the quantity representing the oscillations, $A$ is the amplitude of the oscillations, $\omega$ is the angular frequency, and $\phi$ is a phase constant. The angular frequency is related to the frequency by $\omega = 2\pi f=\frac{2\pi}{T}$ with $T$ the period. The amplitude and phase constant are determined by the initial conditions.

It is helpful in many situations to characterize the oscillatory motion in terms of the frequency.  This can be done by transforming from the *time domain* to the *frequency domain*. So, rather than considering how the oscillatory signal depends on time you examine how it depends on frequency.

## Part 2:  AC Signals and Filters

### Complex Impedance 

The idea of a voltage divider can be used for circuits that include elements besides resistors.  The generic circuit is shown in the figure below.  The input and output voltages depend on the contents of the boxes indicated by $Z_1$ and $Z_2.$  Those can contain resistors, capacitors, and inductors, so the impedances may be complex.



Figure: Voltage divider with complex impedances.

The generalized resistance, or **impedance**, for different resistors, capacitors, and inductors are given by 
$$Z_R=R,\ Z_C=\frac{1}{j\omega C},\ Z_L = j\omega L,$$
where $\omega=2\pi f$ is the angular frequency of an AC signal and we use $j=\sqrt{-1}$ to indicate an imaginary number. Note that the impedance for capacitors and inductors depends on frequency, $\omega=2\pi f$. Following our earlier treatment we can write the gain of a voltage divider, $G$, as
$$G =\frac{V_\text{out}}{V_\text{in}}=\frac{Z_2}{Z_1+Z_2}.$$
This has the same mathematical form as the resistor case where we have replaced the resistors with (possibly) complex impedances. Given that the complex impedances can depend on frequency we may build circuits that behave differently for signals with different frequencies.

Note that $G$ is a complex number.  Any complex number can be written in terms of its magnitude and phase; we would write gain as $G = |G|e^{i\phi}$.  In terms of the measurements you'll take below, $|G| = |V_\text{out}|/|V_\text{in}|$, and $\phi$ is the phase *difference* between $V_\text{out}$ and $V_\text{in}$ (this can be added as a "Measurement" on your oscilloscope). In general, both $|G|$ and $\phi$ depend on frequency $f$.

## High-pass filter

A **high-pass** filter should let signals with high frequencies *pass through* from the input to output.



Figure: High-pass filter using a resistor, $R$, and capacitor, $C$.

For this arrangement the gain of the high-pass filter is given by
$$G_\text{hp}=\frac{R}{\frac{1}{j\omega C}+R}=\frac{j\omega RC}{1+j\omega RC}$$
In the lab we measure real quantities so we need to find the absolute value of the gain, $|G|=\sqrt{GG^*}$, where $G^*$ is the complex conjugate of $G$. The complex conjugate has $j$ replaced by $-j$.  This gives 
$$|G_\text{hp}|=\sqrt{G_\text{hp}G_\text{hp}^*}=\sqrt{\frac{\omega^2 R^2C^2}{1+\omega^2 R^2C^2}}.$$

## Low-pass filter

A **low-pass** filter should let signals with low frequencies *pass through* from the input to output.



Figure: Low-pass filter using a resistor, $R$, and capacitor, $C$.

For this arrangement the gain of the low-pass filter is given by
$$G_\text{lp}=\frac{\frac{1}{j\omega C}}{R+\frac{1}{j\omega C}}=\frac{1}{j\omega RC+1}$$
As with the high-pass result the gain is complex (it has a $j$ in it).  So we need to find the absolute value of the gain. 
$$|G_\text{lp}|=\sqrt{G_\text{lp}G_\text{lp}^*}=\sqrt{\frac{1}{\omega^2 R^2C^2+1}}.$$

### NOTE

A Fourier Transform can be used to visualize the frequency content of a signal. Any function $x(t)$ can be approximated as the sum of many sinusoidal functions:
$$x(t)\approx \text{Re}(\sum_{n=0}^N a_n e^{iω_nt})$$
where there are many $\omega_n=2\pi f_n$ spread over a wide range of frequencies, and each is associated with a complex amplitude $a_n$ (an amplitude and a phase). The process of obtaining all the $a_n$ is the discrete Fourier transform. (Its continuous, exact cousin the Fourier transform is an essential tool of theoretical physics.) The fast Fourier transform (FFT) is a technique for computing many of the coefficients $a_n$ quickly.
