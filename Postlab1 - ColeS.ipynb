## Solving systems of equations symbolically (bonus)

Solving a system of equations like you just did (3 equations and 3 unknowns) can be tedious and prone to mistakes.  We can have a computer algebra package help us do this.

Our circuit analysis gave the following 3 equations for the 3 unknown currents, $I_i$. We have
$$V_1 = I_1R_1-I_3R_2$$
$$V_2 = I_2R_3+I_3R_2$$
$$I_2 = I_1+I_3$$
with $V_1 = 3 \text{ V}$, $V_2 = 6\text{ V}$, $R_1= 6\ \Omega$, $R_2=2\ \Omega$, $R_3=3\ \Omega$.
To use symbolic algebra we need the `sympy` package, which we `import` as shown.

```python
import sympy as sp
V1,V2,I1,I2,I3,R1,R2,R3 = sp.symbols('V1 V2 I1 I2 I3 R1 R2 R3') #define variables
```

To use the symbolic solver we want each equation to be equal to zero
$$V_1 - I_1R_1 + I_3R_2=0$$
$$V_2 - I_2R_3 - I_3R_2=0$$
$$I_2 - I_1 - I_3=0$$
Now we use the `linsolve` command with `linsolve((equations to solve), (variables to find))`.

```python
currents = sp.linsolve((V1-I1*R1+I3*R2,V2-I2*R3-I3*R2, I2-I1-I3),(I1,I2,I3))
print(currents)
```

These are the currents (in order, $I_1$, $I_2$, $I_3$) in symbolic form.  If you want a number then we will substitute in the values with the `subs` command.

```python
currents.subs([(V1,3),(V2,6),(R1,6),(R2,2),(R3,3)])
```
