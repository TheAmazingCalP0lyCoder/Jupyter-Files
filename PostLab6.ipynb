2. Your team in lab used an unloaded voltage divider to (effectively) measure the resistance of your thermistor, but the team next to you used a loaded voltage divider.  Both circuits are shown below; $R_{1} = R_{2} = 1.5\;k\Omega$, and $T_\text{Therm}$ is the resistance of the thermistor.



Plot overlayed graphs of $V_\text{out}$ vs. $R_\text{Therm}$ for both circuits.  Take the range of $R_\text{Therm}$ to correspond to temperatures slightly below room temperature up to just above the boiling point of water.

<span style = 'background-color:yellow'>Answer to Problem 2 below</span> 

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
#Unloaded Circuit
V_in = 1.5
R_1 = 1500
R_tu = np.linspace(1300,50, 105)
```

```python
#G = V_out_u / V_in
#I = V_in / (R_1 + R_tu)
#V_out_u = V_in - I*R_tu
V_out_u = V_in*( 1 - 1 / ( R_1 / R_tu + 1 ) )
```

```python
#Loaded Circuit
V_in = 1.5
R_1 = 1500
R_2 = 1500
R_tl = np.linspace(1300,50, 105)
```

```python
#V_out_l = V_in - I*R_tl
#I = V_in / ( R_1 + ( 1 / ( (1/R_2) + (1/R_tl) ) ) )
V_out_l = V_in*( 1 - R_tl / ( R_1 + ( 1 / ( (1/R_2) + (1/R_tl) ) ) ) )
```

```python
plt.plot(R_tu, V_out_u, 'bo', label = 'Unloaded Circuit')
plt.plot(R_tl, V_out_l, 'ro', label = 'Loaded Circuit')
plt.grid(True)
plt.xlabel('Thermistor Resistance (Ohm)')
plt.ylabel('Voltage Out')
plt.xlim(0,1350)
plt.ylim(0.6,1.5)
plt.legend()
plt.savefig('Postlab6_Ex2_plot_26Feb2026.png')
```
